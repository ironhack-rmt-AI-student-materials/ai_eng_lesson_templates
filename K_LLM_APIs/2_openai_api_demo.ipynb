# OpenAI API Demo

```python
from openai import OpenAI
import os
from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv()) # read local .env file

OPENAI_API_KEY  = os.getenv('OPENAI_API_KEY')
```

```python
client = OpenAI()

response = client.responses.create(
    model="gpt-6-luna",
    input="What's the capital of France"
)

print(response.output_text)
```

<br>

## Main parameters

The `client.responses.create()` method accepts several parameters.

For a basic request, `model` and `input` are enough but there's quite a few other parameters. Here's the most common ones:

- **`model`**: The model to use (for example, `gpt-6-luna`). Models differ in cost, capabilities and supported parameters.
- **`input`**: The prompt or content sent to the model. It can be a string or a structured list of messages.
- **`instructions`**: Optional system-level guidance that sets the model’s role, tone, format, or rules. Unlike input (the task or question itself), instructions define how the model should respond and take priority over input.
- **`max_output_tokens`**: Sets an upper limit on generated tokens, including reasoning tokens.
- **`temperature`**: Controls randomness when supported. Lower values are less random; higher values produce more varied outputs. Some models do not support it.
- **`tools`**: Optional tools the model can use, such as web search or function calling. Your application may need to handle tool calls and provide results.


More info:
- [Responses API reference](https://developers.openai.com/api/reference/python/resources/responses/methods/create)
- [Model documentation](https://developers.openai.com/api/docs/models), for model-specific support.
