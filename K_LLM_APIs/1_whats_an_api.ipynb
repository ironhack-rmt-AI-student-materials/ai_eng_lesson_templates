# What is an API?

An **API (Application Programming Interface)** is a way for one software application to communicate with another — one application sends a request, and the other returns a response.

Think of it like a **waiter in a restaurant**:
- You (the client) place an order (a request).
- The waiter (the API) delivers it to the kitchen (the server).
- The kitchen prepares the food (processes the request).
- The waiter brings the result back (the response).


The most common type you'll encounter is a Web API, which lets applications communicate over the internet using HTTP. Web APIs allow software to access online resources and services—such as retrieving weather data, sending messages, processing payments, or interacting with AI models.

## Key Concepts
- **Endpoint**: a specific URL you call (e.g. `https://api.example.com/users`)
- **Request**: what you send (method, headers, body)
- **Response**: what you get back (usually JSON)
- **Methods**: `GET` (read), `POST` (create), `PUT/PATCH` (update), `DELETE` (remove)
- **Authentication**: API keys, tokens, etc. — proves who's calling

## Example: Ironhack Characters API

An API that allows you to create, read, update, and delete movie characters.

## Base URL
`https://characters-api.edu.ironhack.com`


## Character Properties
   Property   | Type    | Required | Description          |
 |------------|---------|----------|----------------------|
 | `name`     | String  | Yes      | Character's name     |
 | `occupation` | String | Yes      | Character's job/role |
 | `weapon`   | String  | Yes      | Character's weapon   |
 | `debt`     | Boolean | No       | Whether they owe debt|

---

## Endpoints
 | Endpoint          | HTTP Verb | Description                     |
 |-------------------|-----------|---------------------------------|
 | `/characters`     | GET       | List all characters             |
 | `/characters/:id` | GET       | Get details of a specific character |
 | `/characters`     | POST      | Create a new character          |
 | `/characters/:id` | PUT/PATCH | Update a specific character     |
 | `/characters/:id` | DELETE    | Delete a specific character     |

---
**Note:** Replace `:id` with the actual character ID in the endpoint URLs.


```python
#
# Example: sending a GET request to get a list of characters
#

import requests

def display_characters():
    response = requests.get("https://characters-api.edu.ironhack.com/characters")

    print(response.json())
    print("\nNumber of characters in the API...", len(response.json()))

display_characters()
```

```python
#
# Example: sending a POST request to create a new character
#

import requests

new_character = {
    "name": "Arya Stark",
    "occupation": "Assassin",
    "weapon": "Needle",
    "debt": False
}

response = requests.post(
    "https://characters-api.edu.ironhack.com/characters",
    json=new_character
)

print(response.status_code, response.json())
```

```python
#
# we can now double check if the character was created...
#

display_characters()
```

## Why APIs Matter for AI Engineering

Modern AI applications rarely rely on a single model. Instead, they integrate multiple services—LLMs, vector databases, search, speech, vision, authentication, and external business systems—all connected through APIs.

APIs are essential because they enable AI engineers to:

- **Access foundation models** (e.g., OpenAI, Anthropic, Google, Mistral) without managing the underlying infrastructure.
- **Build AI-powered workflows** by connecting models with databases, tools, and third-party services.
- **Leverage the latest capabilities** as providers continuously release new models, reasoning features, multimodal support, and agentic tools through their APIs.
- **Scale reliably** with standardized interfaces for authentication, rate limiting, monitoring, and billing.
- **Stay model-agnostic**, making it easier to switch providers or combine multiple AI services as the ecosystem evolves.

Key takeaway: APIs are the backbone of modern AI engineering—they allow developers to focus on building intelligent applications rather than training and operating large AI models from scratch.
