# Transfer Learning




## What is Transfer Learning


Transfer Learning = Reusing a pretrained model and adapting it to a new task -instead of training a model from scratch. This can be done by freezing the pretrained weights or adapting them.


<br>


## Main categories of Transfer Learning

- **Feature Extraction**
    - Keep the pretrained model fixed and train only a small task-specific head.
    - Typically used when the pretrained features are already good enough for the new task.
    - Useful when data is limited or compute is low.
    - Example: a classifier on top of a pretrained model.
- **Fine-Tuning**
    - Adapt the pretrained model by updating some or all of its weights.
    - Includes both full fine-tuning and parameter-efficient methods (like LoRA).
    - Full fine-tuning: Update all weights in the pretrained model
    - Parameter-Efficient Fine-Tuning (PEFT): Update only a small subset of parameters or add small trainable modules. Includes methods such as LoRA, Adapters, Prompt Tuning, and Prefix Tuning.
